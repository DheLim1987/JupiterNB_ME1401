```python
nums = [10,7,12,4,6,8,2,1,5,9,3]
n = len(nums)
p = n - 1
for k in range(0,n,1):
    for i in range(0,p,1):
        if(int(nums[i])>int(nums[i+1])):
            t = nums[i]
            nums[i] = nums[i+1]
            nums[i+1] = t
        else:
            continue
    p = p-1
print(nums[len(nums) - 2])
```
