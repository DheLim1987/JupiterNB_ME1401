```python
import matplotlib.pyplot as plt
import numpy as np

```

```python
# plt.plot([1,2,3],[2,4,6])
x = [1,2,3,4,5]
y = [2,4,6,7,8]


plt.plot(x,y, label='y=2x', color = 'red', linewidth = 5, marker= '.', markersize = 15)
#plt.plot(x,y,'r.--', label = 'y = 2x


plt.title('Our first graph', fontdict={'fontname': 'Comic Sans MS', 'fontsize': 20})
plt.xlabel('Bananas')
plt.ylabel('Price')

plt.xticks([0,1,2,3])
plt.yticks([2,4,6,8,10])

plt.legend()

plt.show()
```
