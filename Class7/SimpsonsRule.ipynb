```python
import numpy as np
import matplotlib.pyplot as plt
lowerbound = int(input("Lowerbound for the function: "))
upperbound = int(input("Upperbound for the function: "))
n = int(input("Tell Number of small divisions you require: "))
div = (upperbound - lowerbound)/n
area = 0
def f(x):
    y = (x**2)
    return y
i = lowerbound
while(i<=upperbound):
    y0 = f(i - div)
    y1 = f(i)
    y2 = f(i + div)
    area = area + (div/3)*(y0 + (4*y1) + y2)
    i = i+ (2*div)
print(area)
```
