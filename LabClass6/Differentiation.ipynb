```python
#delta x = 0.1
# u = 7 - xtan(x)

import math

x = float(input("Give the value of x in [-1,1]"))
h = 0.1

def u(x):
    u = 7 - x*(math.tan(x))
    return u

def g(x):
    g = - (math.tan(x) + x*((1/math.cos(x))**2))
    return g

#first order:
derivative1 = (u(x+h) - u(x))/h
error1 = derivative1 - g(x)
print(derivative1)
print(error1)


#second order:
derivative2 = (4*(u(x+h)) - u(x+(2*h)) - 3*(u(x)))/(2*h)
error2 = derivative2 - g(x)
print(derivative2)
print(error2)
```

```python
import math
import matplotlib.pyplot as plt
import numpy as np
x = float(input("Give the value of x in [-1,1]"))

def u(x):
    u = 7 - x*(math.tan(x))
    return u

def g(x):
    g = - (math.tan(x) + x*((1/math.cos(x))**2))
    return g

xp = np.array([0.1,0.01,0.001])

#first order:
yp1 = np.zeros(3)
for i in range(0,3,1):
    h = xp[i]
    derivative1 = (u(x+h) - u(x))/h
    error1 = abs(derivative1 - g(x))
    yp1[i] = error1
    print(derivative1)
    print(error1)
#print(yp1)
#fig, ax1 = plt.subplots()
#ax1.plot(xp, yp1)
#ax1.invert_xaxis()
#plt.show()

#second order:
yp2 = np.array([0.0,0.0,0.0])
for i in range(0,3,1):
    h = xp[i]
    derivative2 = (4*(u(x+h)) - u(x+(2*h)) - 3*(u(x)))/(2*h)
    error2 = abs(derivative2 - g(x))
    yp2[i] = error2
    print(derivative2)
    print(error2)
#fig, ax2 = plt.subplots()    
#ax2.plot(xp,yp2)
#ax2.invert_xaxis()
#plt.show()


fig, ax = plt.subplots()

# 2. Plot multiple curves on the SAME axes (ax)
ax.plot(xp, yp1, label='h')
ax.plot(xp, yp2, label='Error')

# 3. Invert the desired axis (e.g., Y-axis)
ax.invert_xaxis()  # Or ax.set_yinverted(True)

# 4. Add formatting (highly recommended for multi-curve plots)
ax.set_title('Bakloli')
ax.set_xlabel('h')
ax.set_ylabel('Error')
ax.legend()  # Displays the labels set in step 2

# 5. Display the plot
plt.show()


```
