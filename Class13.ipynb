```python
import time
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

x = sp.symbols('x')
xs = np.linspace(-2*np.pi, 2*np.pi, 400)
orders = [1,3,5,7,9,13]

polys = {}
for n in orders:
    series_n = sp.series(sp.sin(x), x, 0, n+1)
    polys[n] = series_n.removeO()

t0 = time.perf_counter()
ys_subs = {}
for n in orders:
    values = []
    for v in xs:
        values.append(float(polys[n].subs(x, v)))
    ys_subs[n] = np.array(values)
t_subs = time.perf_counter() - t0
```
