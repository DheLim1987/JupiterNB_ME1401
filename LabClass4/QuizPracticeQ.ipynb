```python
import matplotlib.pyplot as plt
rawData = [1,2,3,4,5]
def main():
    modData = myFunc(rawData, windowSize = 3)
    print(modData)

def myFunc(data, windowSize=3):
    modData=[]
    for i in range(len(data)):
        start = max(0, i - windowSize + 1)
        window = data[start: i+1]
        modData.append(sum(window)/len(window))
    plt.plot(rawData,modData,color='blue')
    plt.show()
    return modData
main()


```
