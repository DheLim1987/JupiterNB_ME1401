```python
l = -1
u = 1
n = int(input("This is the times i want my length to be divided in "))
p = (u-l)/n
Area = 0
x = l
for k in range(1,n,1):
    Area = Area + (1/2)*p*((2*((x)**3)) + (x**2) + (x/2) + 4 + ((2*((x + p)**3)) + ((x+p)**2) + ((x+p)/2)))
    x = x + p
print(Area)
error = 4.666667 - Area
percent = (error/Area)*(100)
print(error)
print(percent)
if(error<0.1):
    print("The value obtained is almost correct.")
else:
    print("Try with higher n value ")
```
