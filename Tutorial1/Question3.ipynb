```python
import math
l = 0.5
u = 2
n = int(input("This is the times i want my length to be divided in "))
a = 0
p = (u - l)/n
x = l
for k in range(1,n,1):
    a = a + (1/2)*(x - (math.log(x)) + (x+p) - (math.log(x+p)))*p
    x = x + p
print(a)
error = 1.6421 - a
percent = (error/a)*(100)
print(error)
print(percent)
if(error<0.1):
    print("The value obtained is almost correct.")
else:
    print("Try with higher n value ")
```
