```python
import math
n = int(input("This is the times i want my length to be divided in "))
x = float(input("The value of x is: "))
area = 0
if(-1 <= x <0):
    u = 0
    l = -1
    p = (u- l)/n
    t = l
    for k in range(1,n,1):
        area = area + (1/2)*(math.sin(t) + math.sin(t+p))*p
        t = t + p
    print(area)
elif(0 <= x < 1):
    u = 1
    l = 0
    p = (u - l)/n
    t = l
    for k in range(1,n,1):
        area = area + (1/2)*((t**3) + ((t+p)**3))*p 
        t = t + p
    print(area)
elif(1 <= x < 6):
    u = 6
    l = 1
    p = (u - l)/n
    t = l
    for k in range(1,n,1):
        area = area + (1/2)*(((math.e)**(1-t)) + ((math.e)**(1-(t+p))))*p
        t = t + p
    print(area)
else:
    print("Invalid Input: ")
```
