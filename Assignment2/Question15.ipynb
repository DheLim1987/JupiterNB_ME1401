```python
import numpy as np
import matplotlib.pyplot as plt

n = [10,100,1000]
lower = float(input("Enter lower limit for integral: "))
upper = float(input("Enter upper limit for integral: "))

def f(x):
    f = (1-(x**2))*(0.5)
    return f

## Trapezoidal Rule:
Area_trap = []
for i in n:
    interval = (upper - lower)/i
    area = 0
    for j in range(0,i,1):
        a = lower + (j*interval)
        b = a + interval
        area = area + ((f(a) + f(b))*(interval/2))
    Area_trap.append(area)

## Simpsons Rule:
Area_sim = []
for i in n:
    interval = (upper - lower)/i
    area = 0
    for j in range(0,i,1):
        a = lower + (j*interval)
        b = a + interval
        area = area + ((interval/6)*(f(a) + (4*f((a+b)/2)) + f(b)))
    Area_sim.append(area)


plt.plot(n, Area_trap, color='red', marker = 'o', label = 'Trapezoidal Rule')
plt.plot(n, Area_sim, color = 'blue', marker = 's', label = 'Simpsons Rule' )
plt.xlabel('Number of subintervals')
plt.ylabel('Area')
plt.title('Comparing Area Accuracy')
plt.legend()
plt.show()
```
