```python
import matplotlib.pyplot as plt
import numpy as np

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
electronics = [1200, 1500, 1700, 1300, 1600, 1800, 2000, 2100, 1900, 2200, 2400, 2500]
clothing = [800, 950, 1100, 1050, 1200, 1500, 1300, 1400, 1350, 1600, 1700, 1750]

index = np.argmax(electronics)          
peak_month = months[index]
peak_value = electronics[index]

fig,ax = plt.subplots(1,2, figsize = (12,10))

ax[0].plot(months,electronics, color = 'red', marker='o', linestyle='-', label='Electronics')
ax[0].plot(months,clothing, color = 'blue', marker='s', linestyle='--', label='Clothing')
ax[0].set_title('Line Chart')
ax[0].set_xlabel('Month')
ax[0].set_ylabel('Sales')
ax[0].legend()
ax[0].grid(True)
ax[0].annotate(f'Highest: {peak_month} ({peak_value})', xy=(peak_month, peak_value), xytext=(index - 3, peak_value + 150), arrowprops=dict(arrowstyle='->', color='black'), fontsize=15)

x = np.arange(len(months))
w = 0.2
ax[1].bar(x - w/2, electronics, width=w, color='red', label='Electronics')
ax[1].bar(x + w/2, clothing, width=w, color='blue', label='Clothing')
ax[1].set_xticks(x)
ax[1].set_xticklabels(months)
ax[1].set_title('Bar Chart')
ax[1].set_xlabel('Month')
ax[1].set_ylabel('Sales')
ax[1].legend()
ax[1].annotate(f'Highest: {peak_month} ({peak_value})', xy=(index - w/2, peak_value), xytext=(index - 4, peak_value + 150), arrowprops=dict(arrowstyle='->', color='black'), fontsize=15)

fig.suptitle('Monthly Sales: Electronics vs Clothing', fontsize=25)
plt.tight_layout()
plt.savefig('sales_plot.png', dpi=300)
plt.show()
```
