```python
#Bisection Method
def f(x):
    output = x**5 - 25500
    return output

lower = 1
upper = 10

for i in range(0,20,1):
    mid = (upper + lower)/2
    if(f(upper)*f(mid) > 0):
        upper = mid
    elif(f(upper)*f(mid) < 0):
        lower = mid
    else:
        break
print(f"The root by Bisection Method is {mid}.")
        
```

```python
#Newton - Raphson Method
def f(x):
    output = x**5 - 25500
    return output

def f_derivative(x):
    output = 5*(x**4)
    return output

def newton_raphson(x):
    output = x - (f(x)/f_derivative(x))
    return output

x0 = 5
for i in range(0,20,1):
    output = newton_raphson(x0)
    x0 = output

print(f"The root by Newton-Raphson Method is {x0}")
```
