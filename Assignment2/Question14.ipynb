```python
import numpy as np

n = 3000
print(f"The number of subintervals used are {n}")
lower = int(input("Tell the lower boundary of integration: "))
upper = int(input("Tell the upper boundary of integration: "))
interval = (upper-lower)/n

def f(x):
    f = float((1/(1+(x**2))))
    return f

Area = 0
for i in range(0,n,1):
    x = (lower + (i*interval))
    Area = Area + (f(x) + f(x+interval))*(interval/2)

Area_real = np.arctan(upper) - np.arctan(lower)

print(Area)
print(Area_real)

tolerance = 10**(-8)
if(abs(Area - Area_real)>tolerance):
    print("Increase the sub intervals")
else:
    print("Choosen the correct number of subintervals")

    
```
