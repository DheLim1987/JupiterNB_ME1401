```python
import time
import numpy as np
import random

##Creating Numpy Random Generator
def numpy_random(n1,n2,l):
    arr = np.random.uniform(n1,n2,l)
    return arr

#Creating Python Random Generator
def python_random(n1,n2,l):
    list1 = []
    for i in range(0,l,1):
        x = random.uniform(n1,n2)
        list1.append(x)
    arr = np.array(list1)
    return arr

#Creating Main function
def main():
    #Taking input
    n1 = float(input("Enter the first number you want the boundary to be set for choosing random numbers: "))
    n2 = float(input("Enter the last number you want the boundary to be set for choosing random numbers: "))
    l = int(input("The number of random numbers to be generated are: "))

    #Using numpy
    start1 = time.perf_counter()
    arr1 = numpy_random(n1,n2,l)
    end1 = time.perf_counter()
    print(arr1)
    print(f"Time taken is: {end1 - start1}")

    #Using random function
    start2 = time.perf_counter()
    arr2 = python_random(n1,n2,l)
    end2 = time.perf_counter()
    print(arr2)
    print(f"Time taken is: {end2 - start2}")


main()
```
