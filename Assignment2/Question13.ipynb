```python
import numpy as np
import math

def f(x,y):
    F = np.array([[y*(math.cos(x*y)) + 1],[math.sin(x*y) + x - y]])
    return F

def jac(x,y):
    jacb = np.array([[(-(y**2)*math.sin(x*y)),(math.cos(x*y) - (x*y*math.sin(x*y)))],[((y*(math.cos(x*y))) + 1), ((x*(math.cos(x*y)))-1)]], dtype = float)
    return jacb

def inverse(jacb):
    n = len(jacb)
    I = np.identity(n)
    jacb1 = np.concatenate((jacb,I),axis = 1,dtype=float)
    #print(jacb1)

    #Calculating lower traingular matrix

    for j in range(0,n-1,1):
        for i in range(j,n-1,1):
            prefac = jacb1[i+1,j]/jacb1[j,j]
            jacb1[(i+1)] = jacb1[(i+1)] - (prefac*(jacb1[j]))
    
    #Calculating Upper triangular matrix

    for j in range(n-1,-1,-1):
        for i in range(0,j,1):
            prefac = jacb1[i,j]/jacb1[j,j]
            jacb1[i] = jacb1[i] - (prefac*(jacb1[j]))

    #Making left side identity

    for i in range(0,n,1):
        jacb1[i] = jacb1[i]/(jacb1[i,i]) 
    #print(jacb1)

    #Calculating Inverse

    inverse = I
    for i in range(0,n,1):
        for j in range(0,n,1):
            inverse[i,j] = jacb1[i,j+n]
    #print(inverse)
    return inverse

def calculating_x(inverse,x,y,F):
    delta = -(np.matmul(inverse,F))
    x = x + delta[0][0]
    y = y + delta[1][0]
    return x,y
```

```python
x,y = 1.0,2.0
for i in range(0,10,1):
    F = f(x,y)
    jacb = jac(x,y)
    inv = inverse(jacb)
    x,y = calculating_x(inv,x,y,F)
output1 = np.array([[x],[y]])
print(f"The required solution is: {output1}")
```
