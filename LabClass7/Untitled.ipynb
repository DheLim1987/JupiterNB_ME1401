```python
import sympy as sp
from IPython.display import display
sp.init_printing()

x,y = sp.symbols('x y')

expr = (x+y)**3
display(sp.expand(expr))
display(sp.factor(x**3 - x))
display(sp.simplify(sp.sin(x)**2 + sp.cos(x)**2))
display(sp.trigsimp(sp.sin(x)/sp.cos(x)))
```

```python
import sympy as sp
from IPython.display import display
sp.init_printing()

x,y = sp.symbols('x y')

eq = sp.Eq(x**2 - 5*x + 6, 0)
display(eq)
display(sp.solve(eq,x))

display(sp.solve(sp.Eq(x**2 + y**2, 1),y))
```

```python
import sympy as sp
from IPython.display import display
sp.init_printing()

x,y = symbols('x y')

eq1 = sp.Eq(x+y, 5)
eq2 = sp.Eq(x-y, 1)
display(sp.solve([eq1, eq2], [x, y]))

display(sp.solve([sp.Eq(x**2 + y + 1, 0), sp.Eq(x-y, 0)], [x,y]))
```

```python
import sympy as sp
from IPython.display import display
sp.init_printing()

x = sp.symbols('x')

display(sp.integrate(sp.sin(x), x))
display(sp.integrate(x**2, (x,0,1)))
display(sp.integrate(sp.exp(-x**2), (x, -sp.oo, sp.oo)))
```

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display
sp.init_printing()

x = sp.symbols('x')
expr = sp.sin(x)*sp.exp(-x)
display(expr)

f = sp.lambdify(x, expr, 'numpy')
xs = np.linspace(0,10,200)
ys = f(xs)
display(ys[:4])

plt.plot(xs,ys)
plt.xlabel('x')
plt.grid(alpha = 0.3)
plt.show()
```

```python
import sympy as sp
from IPython.display import display
sp.init_printing()
import matplotlib.pyplot as plt
import numpy as np

orders = [1,3,5,7,9,11,13,15]

x_vals = np.linspace(-(2*np.pi), 2 * np.pi, 200)
y1 = np.sin(x_vals)

x = sp.symbols('x')
f = sp.sin(x)




for i in orders:
    display(f.series(f,0,orders[i]))
    edit = (f.series(x,0,i+1).removeO())
    display(edit)

    numerical_func = sp.lambdify(x, edit, 'numpy')
    y2 = numerical_func(x_vals)
    plt.plot(x_vals, y2)

plt.plot(x_vals, y1, color="blue")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True, alpha=0.3)

plt.show()
```
