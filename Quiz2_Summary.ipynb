```python
# Solution of Linear Equations using Gauss Elimination:
## Assume a nxn matrix with n systems to solve

import numpy as np

A = np.array([[5,4,3],[3,2,2],[2,1,-2]], dtype = float)   ### Taken values of A and b to check the code
b = np.array([50,25,30], dtype = float)
n = len(A)

# To solve Ax = b

def augmented_matrix(A,b):
    b_transpose = b.reshape(-1,1)
    aug_mat = np.concatenate((A,b_transpose), axis = 1, dtype = float)
    return aug_mat

def forward_substitution(aug_mat):
    for i in range(1,n,1):
        for j in range(0,i,1):
            prefac = aug_mat[i][j]/ aug_mat[j][j]
            aug_mat[i] = aug_mat[i] - prefac* (aug_mat[j])
    return aug_mat

def backward_substitution(aug_mat):
    x = np.zeros(n, dtype = float)  ## To edit value of x after each iteration  
    for i in range(n-1,-1,-1):
        sum = 0
        for j in range(0,n,1):
            sum = sum + (aug_mat[i][j]*x[j])
        x[i] = (aug_mat[i][n] - sum)/ aug_mat[i][i]
    return x

aug_mat = augmented_matrix(A,b)
aug_mat = forward_substitution(aug_mat)
x = backward_substitution(aug_mat)
print(x)
```

```python
# Solution of Linear Equations using LU Decomposition
## Assume a nxn matrix to solve

import numpy as np

A = np.array([[5,4,3],[3,2,2],[2,1,-2]], dtype = float)   ### Taken values of A and b to check the code
b = np.array([50,25,30], dtype = float)
n = len(A)

def triangular(A):
    ### A = LU
    L = np.eye(n)
    U = A
    for i in range(1,n,1):
        for j in range(0,i,1):
            prefac = A[i][j]/A[j][j]
            L[i][j] = prefac
            U[i] = U[i] - prefac* (U[j])
    return L,U

def forward_elimination(L,b):
    ### Ly = b
    y = np.zeros(n)
    for i in range(0,n,1):
        sum = 0
        for j in range(0,n,1):
            sum = sum + (L[i][j]*y[j])
        y[i] = (b[i]-sum)
    return y

def backward_substitution(U,y):
    ### Ux = y
    x = np.zeros(n, dtype = float)  ## To edit value of x after each iteration  
    for i in range(n-1,-1,-1):
        sum = 0
        for j in range(0,n,1):
            sum = sum + (U[i][j]*x[j])
        x[i] = (y[i] - sum)/ U[i][i]
    return x

L,U = triangular(A)
y = forward_elimination(L,b)
x = backward_substitution(U,y)
print(x)
```

```python
# Finding Inverse of a Matrix

import numpy as np

A = np.array([[5,4,3],[3,2,2],[2,1,-2]], dtype = float)

def inverse(A):
    n = len(A)
    I = np.eye(n, dtype = float)
    Combi = np.concatenate((A,I), axis=1, dtype = float)
    for i in range(1,n,1):
        for j in range(0,i,1):
            prefac = Combi[i][j]/Combi[j][j]
            Combi[i] = Combi[i] - prefac*Combi[j]
    for i in range(n-2,-1,-1):
        for j in range(n-1,i,-1):
            prefac = Combi[i][j]/Combi[j][j]
            Combi[i] = Combi[i] - prefac*Combi[j]
    for i in range(0,n,1):
            Combi[i]=Combi[i]/Combi[i][i]
    inv = np.eye(n)
    for i in range(0,n,1):
        for j in range(0,n,1):
            inv[i][j] = Combi[i][j+n]
    return inv

print(inverse(A))
    
```

```python
# Newton - Raphson Method Multiple Equation

import numpy as np
import math

def f(x,y,z):
    F = np.array([[x + y + z -3],[(x**2) + (y**2) + (z**2) - 5],[((math.e)**(x)) + (x*y) - (x*z) - 1]])
    return F

def jac(x,y,z):
    jacb = np.array([[1,1,1],[(2*x),(2*y),(2*z)],[(((math.e)**x) + y - z), x, (-x)]], dtype = float)
    return jacb

def inverse(jacb):
    n = len(jacb)
    I = np.identity(n)
    jacb1 = np.concatenate((jacb,I),axis = 1,dtype=float)
    #print(jacb1)

    #Calculating lower traingular matrix

    for j in range(0,n-1,1):
        for i in range(j,n-1,1):
            prefac = jacb1[i+1,j]/jacb1[j,j]
            jacb1[(i+1)] = jacb1[(i+1)] - (prefac*(jacb1[j]))
    
    #Calculating Upper triangular matrix

    for j in range(n-1,-1,-1):
        for i in range(0,j,1):
            prefac = jacb1[i,j]/jacb1[j,j]
            jacb1[i] = jacb1[i] - (prefac*(jacb1[j]))

    #Making left side identity

    for i in range(0,n,1):
        jacb1[i] = jacb1[i]/(jacb1[i,i]) 
    #print(jacb1)

    #Calculating Inverse

    inverse = I
    for i in range(0,n,1):
        for j in range(0,n,1):
            inverse[i,j] = jacb1[i,j+n]
    #print(inverse)
    return inverse

def calculating_x(inverse,x,y,z,F):
    delta = -(np.matmul(inverse,F))
    x = x + delta[0][0]
    y = y + delta[1][0]
    z = z + delta[2][0]
    return x,y,z
    
x,y,z = 0.1, 1.2, 2.5
for i in range(0,10,1):
    F = f(x,y,z)
    jacb = jac(x,y,z)
    inv = inverse(jacb)
    x,y,z = calculating_x(inv,x,y,z,F)
output1 = np.array([[x],[y],[z]])
print(f"The required solution is: {output1}")

```
