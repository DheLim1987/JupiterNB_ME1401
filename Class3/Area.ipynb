```python
a = float(input("The lower limit is: "))
b = float(input("The upper limit is: "))
n = int(input("The number of division you want to calculate it with: "))
dx = (b - a)/n
x = 0
area = 0
for k in range(1,n,1):
    area = area + (dx)*(((x)**2) + ((x + dx)**2))/2
    x = x + dx
print(area)
error = 0
if(area>=0.3333333):
    error = area - 0.3333333
    print("The error in area is : ", error)
else:
    error = 0.3333333 - area
    print("The error in area is : ", error)
    
    
    
```
