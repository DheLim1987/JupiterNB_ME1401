```python
import numpy as np
import matplotlib.pyplot as plt

def function1(x):
    y = (x**3) - (3*(x**2)) - (3*x) - 4
    return y
    
init_x = 2.5
fin_x = 5
mid_x = 0

for i in range(0,100,1):
    mid_x = (init_x + fin_x)/2
    y1 = function1(init_x)
    y2 = function1(fin_x)
    y3 = function1(mid_x)
    if((y1*y3)>0):
        init_x = mid_x
    elif(y3 == 0):
        break
    else:
        fin_x = mid_x
    
print(mid_x)
```
