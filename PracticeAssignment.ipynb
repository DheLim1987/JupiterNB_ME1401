```python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt

x,y,z = sp.symbols('x y z')
f = x**2 + y**2 + z**2 - 14
g = x*y + z - 5
h = sp.exp(x-1) + y - z

f_x = sp.diff(f,x)
f_y = sp.diff(f,y)
f_z = sp.diff(f,z)
g_x = sp.diff(g,x)
g_y = sp.diff(g,y)
g_z = sp.diff(g,z)
h_x = sp.diff(h,x)
h_y = sp.diff(h,y)
h_z = sp.diff(h,z)

Jacobian = sp.Matrix([[f_x, f_y, f_z],[g_x, g_y, g_z],[h_x, h_y, h_z]])
Jacobian_func = sp.lambdify((x,y,z), Jacobian, 'numpy')
b = sp.Matrix([[f],[g],[h]])
b_func = sp.lambdify((x,y,z), b, 'numpy')

mat1 = np.array([[1.5],[1.5],[2.5]])
mat2 = np.array([[2],[1],[2]])

def newton_raphson(mat):
    delta = np.linalg.solve(Jacobian_func(mat[0][0],mat[1][0],mat[2][0]),b_func(mat[0][0],mat[1][0],mat[2][0]))
    mat = mat - delta
    return mat

for i in range(0,5,1):
    mat1 = newton_raphson(mat1)
print(mat1)

for i in range(0,5,1):
    mat2 = newton_raphson(mat2)
print(mat2)
```
