```python
import math
import matplotlib.pyplot as plt
n = int(input("Enter the final limit you want to have"))
div = int(input("Enter the divisions to decide your accuracy"))
p = 1/div
x = []
y = []
k = 0.1
while(k<=n):
    x.append(k)
    y.append(math.sin(k)/k)
    k += p
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Sample')
plt.plot(x,y,linestyle='-', color = 'red', linewidth = 2)
plt.show()

```
