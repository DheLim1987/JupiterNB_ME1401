```python
import math
p = math.pi + 0.0000001
q = math.pi - 0.0000001
n = int(input("The number of divisions you want your area to be: "))
div = 1/n
a = ((math.pi)**3 - (p**3))/n
b = ((math.pi)**3 - (q**3))/n
area = 0
i = p
for k in range(0,n,1):
    area = area + math.cos((i)**(1/3))
    i = i + a
limit1 = area/((p - (math.pi/2))**2)
t = q
for k in range(0,n,1):
    area = area + math.cos((t)**(1/3))
    t = t + b
limit2 = area/((p - (math.pi/2))**2)
if(limit1==limit2):
    print("Limit Exists")
    print(f"Limit is: {limit1}")
else:
    print("Limit Does not exist.")
    print(f"Values obtained are {limit1} and {limit2}")
```
