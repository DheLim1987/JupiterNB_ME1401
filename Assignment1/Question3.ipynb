```python
import math
x = int(input("The number you want to make divisions into: "))
p = (math.pi)/(2*x)
area = 0
n = 0
for k in range(0,x,1):
    area = area + (math.sin(n) /(math.sin(n) + math.cos(n)))*p
    n = n + p
print(f"The required area is {area}")
if(area>=(math.pi/4)):
    error = ((area - (math.pi/4))*100)/area
else:
    error = (((math.pi/4) - area)*100)/area
if(error<0.1):
    print(f"The error value is {error}")
else:
    print(f"The error value is {error}, increase value of number of divisions")
```
