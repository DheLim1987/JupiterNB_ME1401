```python
print("ax^2 + bx + c = 0")
a = float(input("The coefficient of x^2 is: "))
b = float(input("The coefficient of x is: "))
c = float(input("The constant term is: "))
result1 = (-b + (((b**2) - (4*a*c))**(1/2)))/(2*a)
result2 = (-b - (((b**2) - (4*a*c))**(1/2)))/(2*a)
print(f"The required values are {result1} and {result2}.")
```
