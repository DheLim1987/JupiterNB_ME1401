```python
import math
def integral(x):
    lower = x ** 3
    upper = (math.pi / 2) ** 3
    n = 1000
    h = (upper - lower) / n
    area = 0
    t = lower
    LHL = 0
    RHL = 0
    for i in range(n):
        middle = t + h / 2
        area = area + math.cos(middle ** (1 / 3)) * h
        t = t + h
    return area
dx = 0.1
for i in range(0,6,1):
    dx = dx / 10
    x = math.pi/2 - dx
    div = (x - math.pi/2) ** 2
    LHL = integral(x) / div
dx = 0.1
for i in range(0,6,1):
    dx = dx / 10
    x = math.pi/2 + dx
    div = (x - math.pi/2) ** 2
    RHL = integral(x) / div
if(LHL-RHL<=0.00001):
    print(f"Limit exists and value is {LHL}")
else:
    print(f"Limit does not exist and value of LHL is {LHL} and value of RHL is {RHL}")
```
