```python
print("Enter the coefficients of cubic equation in form ax^3 + bx^2 + cx + d = 0")
a = float(input("The value of a is: "))
b = float(input("The value of b is: "))
c = float(input("The value of c is: "))
d = float(input("The value of d is: "))
x = 2.5
f = ((-d - b*(x**2) - (c*x))/a)**(1/3)
i = 0
while(abs(f-x)>0.00001):
    x = f
    f = ((-d - b*(x**2) - (c*x))/a)**(1/3)
    i = i+1
print(f"The closest root is: {x}")
print(i)
```
