```python
import numpy as np
import matplotlib.pyplot as plt

Force = np.array([14,18,8,9,13])
Displacement = np.array([0.013,0.020,0.009,0.010,0.012])

spring_constant = Force/Displacement
potential_energy = (0.5)*(spring_constant)*(Displacement**2)

max = potential_energy[0]
min = potential_energy[0]
for i in potential_energy:
    if (i>max):
        max = i
    elif(i<min):
        min = i
    else:
        pass
print(f"The maximum value is {max} and the minimum value is {min}")


springs = np.array([1,2,3,4,5])

plt.bar(springs,spring_constant, width = 0.5, color='red', label = 'Spring_Constant')
plt.xlabel("Springs")
plt.ylabel("Spring_Constants")
plt.title("Spring_Constant vs Springs")
plt.legend()
plt.show()
plt.savefig("Spring_Constants.png")
```

```python
# log 10 (x**2 + y) -1 + y = 0
# root x + xy = 0

import numpy as np
import sympy as sp

x,y = sp.symbols('x y')

f = sp.log(x**2 + y, 10) - 1 + y
g = sp.sqrt(x) + x*y

f_x = sp.diff(f,x)
f_y = sp.diff(f,y)
g_x = sp.diff(g,x)
g_y = sp.diff(g,y)

A = sp.Matrix([[f_x,f_y],[g_x,g_y]])
A_func = sp.lambdify((x,y), A, 'numpy')

b = sp.Matrix([[f],[g]])
b_func = sp.lambdify((x,y), b, 'numpy')

mat = np.array([[2],[-0.5]], dtype = float)

def newton_raphson(mat):
    delta = np.linalg.solve(A_func(mat[0][0],mat[1][0]),b_func(mat[0][0],mat[1][0]))
    mat = mat - delta
    return mat

for i in range(0,100,1):
    mat = newton_raphson(mat)
print(np.round(mat,6))
```
