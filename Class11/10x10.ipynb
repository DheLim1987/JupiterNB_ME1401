```python
import numpy as np
mat1 = np.random.rand(10,10)
mat2 = np.random.rand(10,1)
aug_mat = np.concatenate((mat1,mat2),axis = 1,dtype="float")
print(aug_mat)
for j in range(0,9,1):
    for i in range(j,9,1):
        prefac = aug_mat[(i+1),j]/aug_mat[j,j]
        aug_mat[(i+1)] = aug_mat[(i+1)] - (prefac*(aug_mat[j]))
print(aug_mat)

np.linalg.solve(mat1,mat2)
```
