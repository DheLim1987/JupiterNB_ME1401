```python
import matplotlib.pyplot as plt


# x**3 - 3*(x**2) - 3*x - 4 = 0
x = 2.5
def f(a):
    res = (a**3) - (3*(a**2)) - (3*a) - 4
    return res
def g(a):
    res = (3*(a**2)) - (6*a) - 3
    return res

lis1 = []
lis2 = []

for i in range(0,100,1):
    x = x - (f(x)/g(x))
    lis1.append(x)
    lis2.append(f(x))

plt.plot(lis1,lis2)
plt.show()
print(x)
```
