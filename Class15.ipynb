```python
# Solve using N-R Method
# 1 + z**2 = 0
# z0 = 1 + i/2


def f(x):
    result = 1 + (x**2)
    return result

def f_derivative(x):
    result = 2*x
    return result

def NR_Method(x):
    x = x - (f(x)/f_derivative(x))
    return x

x = 1 + 0.5j
for i in range(0,8,1):
    x = NR_Method(x)
print(x)
```

```python
#Compute A^(-1)

import numpy as np
import sympy as sp

A = np.array([[1,2,0,0],[0,1,3,0],[0,0,1,4],[0,0,0,1]])
I = np.eye(4)
A_new = np.concatenate((A,I),axis = 1,dtype=float)

for i in range(2,-1,-1):
    prefac = A_new[i][i+1]
    A_new[i] = A_new[i] - prefac*(A_new[i+1])
print(A_new)

inv = np.eye(4)
for i in range(0,4,1):
    for j in range(0,4,1):
        inv[i][j] = A_new[i][j+4]
print(inv)

np_inv = np.linalg.inv(A)
print(np_inv)

A_sp = sp.Matrix([[1,2,0,0],[0,1,3,0],[0,0,1,4],[0,0,0,1]])
print(A_sp.inv())
```
